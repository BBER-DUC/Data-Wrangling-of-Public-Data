# Cloning Hugging Face Data Repo


Cloning a Hugging Face dataset repository pulls the complete set of underlying data files, metadata dictionaries, and configuration scripts directly into your local or cloud environment using Git and Git LFS. This provides a self-contained local copy of the raw files, enabling you to inspect structures, run offline data-wrangling scripts, or query Parquet files directly via tools like DuckDB without network latency. Ultimately, having the physical files on disk gives you full control over partitioning, versioning, and custom pipeline integrations for your machine learning workflows.
