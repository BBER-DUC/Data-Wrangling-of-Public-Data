## Query a remote large CSV



```
csv_url = 'https://huggingface.co/datasets/rwcitek/medicare/resolve/main/DAC_NationalDownloadableFile.csv'
csv_url
```

```
# Read, apply schema, convert to parquet
duckdb.query(f"""
  select count(1) as count
  from "{csv_url}"
""").df().style.format('{:,}')
```

## Querying remote parquet

```
parquet_url = 'https://huggingface.co/datasets/rwcitek/medicare/resolve/main/DAC_NationalDownloadableFile.parquet'
parquet_url
```

```
# Read, apply schema, convert to parquet
duckdb.query(f"""
  select count(1) as count
  from "{parquet_url}"
  ;
""").df().style.format('{:,}')
```

```
# Read, apply schema, convert to parquet
duckdb.query(f"""
  select *
  from "{parquet_url}"
  where state = 'NM'
  ;
""").df()
```
